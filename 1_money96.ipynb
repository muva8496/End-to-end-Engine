write the validation rules

Definition: a validation rule is a yes/no test every row must pass. A row that fails gets a reason, so it can be quarantined instead of silently deleted. Objective: one function that splits the data into clean rows and rejected rows, with a reason on each rejected one. Real-world use: when the finance team asks why 12 transactions are missing from the report, the answer is in the rejects table, not in somebody’s memory.

Two design choices:

First failing rule wins. Each rejected row gets exactly one reason.
Nothing disappears. Clean rows plus rejected rows must always equal the rows received. We’ll test that.

```python
def validate(df):
    rules = [
        ("missing_transaction_id", df["transaction_id"].isna()),
        ("duplicate_transaction_id", df["transaction_id"].duplicated(keep="first")),
        ("invalid_date", df["sale_date"].isna()),
        ("invalid_quantity", df["quantity"].isna() | (df["quantity"] < 1)),
        ("negative_price_or_cost", (df["unit_price_kes"] < 0) | (df["unit_cost_kes"] < 0)),
        ("total_sales_mismatch", df["quantity"] * df["unit_price_kes"] != df["total_sales_kes"]),
        ("profit_mismatch", df["total_sales_kes"] - df["total_cost_kes"] != df["profit_kes"]),
    ]
    reason = pd.Series("", index=df.index)
    for name, failed in rules:
        reason = reason.mask((reason == "") & failed, name)

    clean = df[reason == ""]
    rejects = df[reason != ""].assign(reject_reason=reason[reason != ""])
    return clean, rejects
```

```python
clean, rejects = validate(raw_df)
print("received:", len(raw_df), "| clean:", len(clean), "| rejected:", len(rejects))
print("nothing lost:", len(clean) + len(rejects) == len(raw_df))
```

store the data in SQL

Definition: a table with constraints makes the database itself refuse bad data, as a second line of defense behind Pandas. Objective: create two tables, sales_clean for good rows and sales_rejected for quarantined ones, and load the data. Real-world use: the validator is the first guard and the database constraint is the last. If a bug ever slips past the validator, UNIQUE and CHECK still stop the damage.

```python
import sqlite3

con = sqlite3.connect("data/pipeline.db")
con.executescript("""
DROP TABLE IF EXISTS sales_clean;
DROP TABLE IF EXISTS sales_rejected;

CREATE TABLE sales_clean (
    transaction_id     INTEGER PRIMARY KEY,
    sale_date          TEXT    NOT NULL,
    city               TEXT,
    branch             TEXT,
    category           TEXT,
    product            TEXT,
    quantity           INTEGER NOT NULL CHECK (quantity >= 1),
    unit_price_kes     INTEGER NOT NULL CHECK (unit_price_kes >= 0),
    unit_cost_kes      INTEGER NOT NULL CHECK (unit_cost_kes >= 0),
    total_sales_kes    INTEGER NOT NULL,
    total_cost_kes     INTEGER NOT NULL,
    profit_kes         INTEGER NOT NULL,
    sales_channel      TEXT,
    payment_method     TEXT,
    customer_gender    TEXT,
    customer_age_group TEXT,
    salesperson        TEXT
);

CREATE TABLE sales_rejected (
    id            INTEGER PRIMARY KEY AUTOINCREMENT,
    reject_reason TEXT NOT NULL,
    raw_json      TEXT NOT NULL
);
""")
```

```python
def store(con, clean, rejects):
    out = clean.copy()
    out["transaction_id"] = out["transaction_id"].astype(int)
    out["sale_date"] = out["sale_date"].dt.strftime("%Y-%m-%d")
    out.to_sql("sales_clean", con, if_exists="append", index=False)

    if len(rejects):
        rej = pd.DataFrame({
            "reject_reason": rejects["reject_reason"].values,
            "raw_json": [r.to_json(date_format="iso") for _, r in rejects.drop(columns="reject_reason").iterrows()],
        })
        rej.to_sql("sales_rejected", con, if_exists="append", index=False)

store(con, clean, rejects)
```

```python
print(pd.read_sql("SELECT COUNT(*) AS rows_stored, SUM(total_sales_kes) AS revenue_kes FROM sales_clean", con))
print("source revenue:", raw_df["total_sales_kes"].sum())
```
